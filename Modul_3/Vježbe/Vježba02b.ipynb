```python
import pandas as pd
import numpy as np
from sklearn.model_selection import ShuffleSplit, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
```

### Preuzmite dataset sa linka koji vam je dostupan u datoteci datasets_klasifikacija_binarni_p4.txt. Morat ćete se prijaviti sa Google računom na Kaggle poslužitelj, da bi mogli preuzeti dataset. Proučite dataset i definirajte što je target varijabla. Ako je ne pronalazite, izaberite neku varijablu za koju postoje samo dvije vrijednosti u stupcu. Iz dataseta izbacite sve značajke koje nisu numeričke, a ako negdje nedostaje vrijednost umetnite medijan. Koristeći metodu cross_val_score (za cv koristite ShuffleSplit(), a testna veličina neka bude 20%) usporedite vrijednosti "roc_auc"  algoritama stabla odlučivanja i logističke regresije, pa ispišite dobivene nizove "roc_auc". Radi se o površini ispod krivulje klasifikatora i te vrijednosti bi trebale biti veće od 0,5. Na kraju izračunajte prosjeke površina ispod krivulje za oba modela i u komentar navedite koji je model bolji.

```python
df = pd.read_csv('/content/Car_Insurance_Claim.csv')
df
```

```python
# target varijabla
y = df["OUTCOME"]

X = df.drop(columns=[ "ID", "OUTCOME", "AGE", "GENDER", "RACE", "DRIVING_EXPERIENCE", "EDUCATION", "INCOME","VEHICLE_YEAR", "VEHICLE_TYPE" ])

X = X.fillna(X.median()) # znam iz prethodnog zadatka koje kolone treba izbaciti te da imam praznih ćelija


dt = DecisionTreeClassifier(random_state=42)
log_reg = LogisticRegression(max_iter=5000, random_state=42)

cv = ShuffleSplit(n_splits=10, test_size=0.2, random_state=42)

dt_scores = cross_val_score(dt, X, y, cv=cv, scoring='roc_auc')
log_reg_scores = cross_val_score(log_reg, X, y, cv=cv, scoring='roc_auc')

print("Decision Tree AUC scores:", dt_scores.mean())
print("Logistic Regression AUC scores:", log_reg_scores.mean())

```

## Zaključak: Oba score-a su iznad 0.5, ali je Logistic Regression bolja u ovom slučaju jer joj score iznosi 0.8428 u odnasu na Decision Tree čiji score iznosi 0.6998
