## Dataset: This synthetic dataset simulates various demographic and socioeconomic factors that influence annual household income. It can be used for exploratory data analysis, predictive modeling, and understanding the relationships between different features and income levels.

```python
df = pd.read_csv("/content/data.csv")
df.head()
```

```python
df.info()
```

```python
# target
y = df["Income"]

X = df.select_dtypes(include=[np.number]).drop(columns=["Income"])


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21
)

lr = LinearRegression()

loocv = LeaveOneOut()
scores = cross_val_score(lr, X_train, y_train, cv=loocv, scoring='neg_mean_squared_error')

rmse_train = np.mean(np.sqrt(-scores))


lr.fit(X_train, y_train)
predicted = lr.predict(X_test)
rmse_test = np.sqrt(metrics.mean_squared_error(y_test, predicted))

print(rmse_test)
print(rmse_train)
```
