```python
threshold = df['squareMeters'].median()

df['squareMeters_cat'] = np.where(df['squareMeters'] < threshold, 'mala', 'velika')


# OneHot encoding
encoder = OneHotEncoder(drop='first', sparse_output=False)
encoded = encoder.fit_transform(df[['squareMeters_cat']])
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['squareMeters_cat']))

df = pd.concat([df.drop(columns=['squareMeters_cat']), encoded_df], axis=1)


#target
y = df["price"]
X = df.drop(columns=['price'])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=21)

#models
lr = LinearRegression()
lasso_reg = linear_model.Lasso()
ridge_reg = linear_model.Ridge()

#LR
lr.fit(X_train, y_train)
predicted_lr = lr.predict(X_test)
rmse_test_lr = np.sqrt(metrics.mean_squared_error(y_test, predicted_lr))
print(rmse_test_lr)

#LASSO
print("-------------------------------------------")
lasso_reg.fit(X_train, y_train)
predicted_lasso = lasso_reg.predict(X_test)
rmse_test_lasso = np.sqrt(metrics.mean_squared_error(y_test, predicted_lasso))
print(rmse_test_lasso)

#RIDGE
print("-------------------------------------------")
ridge_reg.fit(X_train, y_train)
predicted_ridge = ridge_reg.predict(X_test)
rmse_test_ridge = np.sqrt(metrics.mean_squared_error(y_test, predicted_ridge))
print(rmse_test_ridge)


```
