```python
df = pd.read_csv("/content/Indian Liver Patient Dataset (ILPD).csv")
df.info()
```

```python
Y = df['Gender']
X = df.select_dtypes(include=[np.number])

# Imam neke prazne ćelije na "A/G Ratio" koje popunjavam medianom inače mi vraća error
imputer = SimpleImputer(strategy="median")
X_imputed = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)


X_train, X_test, y_train, y_test = train_test_split(
    X_imputed, Y, test_size=0.2, random_state=42, stratify=Y)

nb=naive_bayes.GaussianNB()
scores = cross_val_score(nb, X_train, y_train, cv=5, scoring='accuracy')


fit=nb.fit(X_train, y_train)
preds=fit.predict(X_test)
print(metrics.accuracy_score(y_test, preds))
```
