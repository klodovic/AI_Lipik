```python
#target
y = df["Price"]
# dropping columns
X = df.drop(columns=[ "Car_ID", "Brand", "Model", "Fuel_Type", "Transmission", "Owner_Type"])
# data split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=21)

#models
lasso_reg = linear_model.Lasso()
ridge_reg = linear_model.Ridge()
svr_reg = SVR()
nusvr_reg = NuSVR()
tree_reg = DecisionTreeRegressor()

# Lasso
lasso_pipe = Pipeline([("scaler", StandardScaler()), ("model", linear_model.Lasso())])
lasso_cv_scores = cross_val_score(lasso_pipe, X_train, y_train, cv=10, scoring="neg_root_mean_squared_error")
lasso_cv_rmse_mean = -lasso_cv_scores.mean()
print(f"Lasso:   {lasso_cv_rmse_mean}")

lasso_pipe.fit(X_train, y_train)
lasso_test_pred = lasso_pipe.predict(X_test)
lasso_test_rmse = np.sqrt(np.mean((lasso_test_pred - y_test)**2))
print(f"Lasso test RMSE:   {lasso_test_rmse}")
print("-----------------------------------------------------------------------------------------------------")

# Ridge
ridge_pipe = Pipeline([("scaler", StandardScaler()), ("model", linear_model.Ridge())])
ridge_cv_scores = cross_val_score(ridge_pipe, X_train, y_train, cv=10, scoring="neg_root_mean_squared_error")
ridge_cv_rmse_mean = -ridge_cv_scores.mean()
print(f"Ridge:   {ridge_cv_rmse_mean}")

ridge_pipe.fit(X_train, y_train)
ridge_test_pred = ridge_pipe.predict(X_test)
ridge_test_rmse = np.sqrt(np.mean((ridge_test_pred - y_test)**2))
print(f"Ridge test RMSE:   {ridge_test_rmse}")
print("-----------------------------------------------------------------------------------------------------")

# SVG
svr_pipe = Pipeline([("scaler", StandardScaler()), ("model", SVR())])
svr_cv_scores = cross_val_score(svr_pipe, X_train, y_train, cv=10, scoring="neg_root_mean_squared_error")
svr_cv_rmse_mean = -svr_cv_scores.mean()
print(f"SVR:   {svr_cv_rmse_mean}")

svr_pipe.fit(X_train, y_train)
svr_test_pred = svr_pipe.predict(X_test)
svr_test_rmse = np.sqrt(np.mean((svr_test_pred - y_test)**2))
print(f"SVR test RMSE:   {svr_test_rmse}")
print("-----------------------------------------------------------------------------------------------------")

# NuSVR
nusvr_pipe = Pipeline([("scaler", StandardScaler()), ("model", NuSVR())])
nusvr_cv_scores = cross_val_score(nusvr_pipe, X_train, y_train, cv=10, scoring="neg_root_mean_squared_error")
nusvr_cv_rmse_mean = -nusvr_cv_scores.mean()
print(f"NuSVR:   {nusvr_cv_rmse_mean}")

nusvr_pipe.fit(X_train, y_train)
nusvr_test_pred = nusvr_pipe.predict(X_test)
nusvr_test_rmse = np.sqrt(np.mean((nusvr_test_pred - y_test)**2))
print(f"NuSVR test RMSE:   {nusvr_test_rmse}")
print("-----------------------------------------------------------------------------------------------------")

# Tree
tree_pipe = Pipeline([("model", DecisionTreeRegressor())])
tree_cv_scores = cross_val_score(tree_pipe, X_train, y_train, cv=10, scoring="neg_root_mean_squared_error")
tree_cv_rmse_mean = -tree_cv_scores.mean()
print(f"Tree:   {tree_cv_rmse_mean}")

tree_pipe.fit(X_train, y_train)
tree_test_pred = tree_pipe.predict(X_test)
tree_test_rmse = np.sqrt(np.mean((tree_test_pred - y_test)**2))
print(f"Tree test RMSE:   {tree_test_rmse}")

```
