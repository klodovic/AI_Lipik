### Uz prvo slovo vašeg imena (veliko tiskano, bez "kvačica") zamislite si još jedno slovo po želji koje se značajno vizualno razlikuje (veliko tiskano, bez "kvačica"). Kad budete birali slova držite se predloška sa slike abeceda_5x5_sva_slova.png. Nakon što odlučite preuzmite ta dva slova iz mape POJEDINACNA SLOVA. Obje slike treba konvertirati i to: CROP i RESIZE na veličinu 5x5 pixela sa servisom https://pinetools.com/c-images/. Nakon toga učitajte obje slike u COLAB i napišite algoritam koji će odlučiti koje je od učitana dva slova prvo slovo vašeg imena.  Uočite da ne treba trenirati nikakav model, nego problem rješavamo algoritamski i odlučujemo na osnovu nijanse sive pixela koji sami izaberete, a to će biti onaj čija svjetlina se znatno razlikuje kod izabrana dva slova. Srednji pixel je na lokaciji [2,2]. Algoritam i slike predajte.

```python
image_one = Image.open("picture.png").convert("L") #metoda za nijanse sive
im_one = image_one.resize((5,5), Image.BILINEAR)
plt.figure(figsize=(3, 3))
plt.imshow(im_one, cmap="gray")
```

```python
image_two = Image.open("picture2.png").convert("L") #metoda za nijanse sive
im_two = image_two.resize((5,5), Image.BILINEAR)
plt.figure(figsize=(3, 3))
plt.imshow(im_two, cmap="gray")
```

```python
im_one_array = np.array(im_one)
im_two_array = np.array(im_two)

center_pixel_one = im_one_array[2, 2]
center_pixel_two = im_two_array[2, 2]

if center_pixel_one > center_pixel_two:
    print("Prvo slovo je J")
    plt.imshow(im_one, cmap="gray")
else:
    print("Prvo slovo je K")
    plt.imshow(im_two, cmap="gray")
```
