```python
import numpy as np
jedinice_vektor = np.arange(5)
print(jedinice_vektor, "-->", jedinice_vektor * jedinice_vektor)
print("Umnožak:", np.dot(jedinice_vektor, jedinice_vektor))
```

```python
vektor_red = np.arange(5).reshape(1,5)
vektor_stupac = np.arange(0, 50, 10).reshape(5, 1)
print("Vektor red:", vektor_red,
  "Vektor stupac:", vektor_stupac,
  "Umnožak 1:", np.dot(vektor_red, vektor_stupac), sep='\n')
```

```python
out = np.dot(vektor_stupac, vektor_red)
print("Umnožak 2:\n",out)
```
