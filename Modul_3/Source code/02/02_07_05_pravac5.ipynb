```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

brojljudi = np.arange(1, 11)
brojpiva = np.arange(0, 20) # 0-19 rootbeers
print(brojljudi)
print("--------------------------")
print(brojpiva)
print("--------------------------")
# numpy tool to get cross-product of values (each against each)
# in two paired arrays. try it out: np.meshgrid([0, 1], [10, 20])
# "perfect" for functions of multiple variables
brojljudi, brojpiva = np.meshgrid(brojljudi, brojpiva)
ukupno = 40 * brojljudi + 5 * brojpiva + 20
print(brojljudi)
print("--------------------------")
print(brojpiva)
print("--------------------------")
print(ukupno)

```

```python
# import needed for 'projection':'3d'
from mpl_toolkits.mplot3d import Axes3D
fig,axes = plt.subplots(2, 3,
subplot_kw={'projection':'3d'},
figsize=(9, 6))
kutovi = [0, 45, 90, 135, 180]
for ax,angle in zip(axes.flat, kutovi):
  ax.plot_surface(brojljudi, brojpiva, ukupno)
  ax.set_xlabel("Ljudi")
  ax.set_ylabel("Piva")
  ax.set_zlabel("Ukupno")
  ax.azim = angle
# we don't use the last axis
axes.flat[-1].axis('off')
fig.tight_layout()
```
