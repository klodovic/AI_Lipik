```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as ss
b = ss.distributions.binom
for bacanja in [5, 10, 20, 40, 80]:
  uspjeh = np.arange(bacanja)
  distribucija = b.pmf(uspjeh, bacanja, .5)
  plt.hist(uspjeh, bacanja, weights=distribucija)
plt.xlim(0, 60);
```
