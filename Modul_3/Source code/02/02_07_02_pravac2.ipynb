```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
ljudi = np.arange(1, 11)
trosak = 40.0 * ljudi + 20.0
display(pd.DataFrame({'trosak':trosak.astype(np.int64)}, index=ljudi).T)
ax = plt.gca()
ax.plot(ljudi, trosak, 'bo')
ax.set_ylabel("Total Cost")
ax.set_xlabel("People")
```
