```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso, Ridge, LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import OneHotEncoder
df = pd.read_csv("DATA.csv")
target_col = "CO2"
if 'Unnamed: 5' in df.columns:
    df.drop(columns=['Unnamed: 5'], inplace=True)
df.drop(columns=["Model"], inplace=True)
ohe = OneHotEncoder(drop='first', sparse_output=False)
Z = ohe.fit_transform(df[['Car']])
dummy_cols = list(ohe.get_feature_names_out(['Car']))
df[dummy_cols] = Z
df.drop(columns=['Car'], inplace=True)
numeric_df = df.copy()
X = numeric_df.drop(columns=[target_col])
y = numeric_df[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21)

imputer = SimpleImputer(strategy="median")
X_train_imp = imputer.fit_transform(X_train)
X_test_imp  = imputer.transform(X_test)
print(X_train_imp)
lr_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())])
lr_cv = cross_val_score(
    lr_pipe, X_train_imp, y_train, cv=10, scoring="neg_root_mean_squared_error")
lr_rmse_cv_mean = -lr_cv.mean()
lr_pipe.fit(X_train_imp, y_train)
lr_pred = lr_pipe.predict(X_test_imp)
lr_rmse_test = np.sqrt(mean_squared_error(y_test, lr_pred))
print("\n[LR]   CV RMSE (mean): %.4f" % lr_rmse_cv_mean)
print("[LR]   Test RMSE:      %.4f" % lr_rmse_test)

```
