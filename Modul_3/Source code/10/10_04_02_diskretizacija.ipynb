```python
import pandas as pd
from sklearn import datasets
from sklearn import preprocessing as skpre
import numpy as np

iris = datasets.load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)

column = iris_df[['sepal length (cm)']]
col_mean = column.mean().values

both = column.copy()
both['> Mean'] = skpre.binarize(column, threshold=col_mean[0]).astype(np.bool_)

print('Column Mean:', col_mean)
display(both.iloc[[0,50,100]])
```

```python
sep_len_series = iris_df['sepal length (cm)']
breaks = [sep_len_series.mean(),
          sep_len_series.max()]

print(pd.cut(sep_len_series, breaks).cat.codes[[0, 50, 100]])
```
