```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np
import seaborn as sns

# Učitaj Diamonds (u seabornu)
df = sns.load_dataset("diamonds").dropna()
print(df.describe(include='all'))
#  df ima numeričke + kateg. stupce, target 'price'
num_cols = ["carat","depth","table","x","y","z"]
cat_cols = ["cut","color","clarity"]

X1 = df[num_cols]
y1 = df["price"]

X1_tr, X1_te, y1_tr, y1_te = train_test_split(X1, y1, test_size=0.2, random_state=42)

model1 = LinearRegression()
model1.fit(X1_tr, y1_tr)
pred = model1.predict(X1_te)
print("R² numeričke:", r2_score(y1_te, pred))
print("RMSE numeričke:", np.sqrt(np.mean((y1_te - pred)**2)))

pre = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first", handle_unknown="ignore"), cat_cols),
        ("num", "passthrough", num_cols),
    ],
    remainder="drop")
#handle_unknown="ignore" - ignorira ako se u testnom skupu pojavi nepoznata kategorija
#remainder="drop" - sve stupce koji ne pripadaju kategorijama cat i num briši

model2 = Pipeline([
    ("pre", pre),
    ("lin", LinearRegression())])

X2 = df[cat_cols + num_cols]
y2 = df["price"]

X2_tr, X2_te, y2_tr, y2_te = train_test_split(X2, y, test_size=0.2, random_state=42)

model2.fit(X2_tr, y2_tr)
pred2 = model2.predict(X2_te)
print("R² sve:", r2_score(y2_te, pred2))
print("RMSE sve:", np.sqrt(np.mean((y2_te - pred2)**2)))
```
