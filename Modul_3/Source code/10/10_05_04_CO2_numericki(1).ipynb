```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso, Ridge, LinearRegression
from sklearn.svm import SVR, NuSVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
df = pd.read_csv("DATA.csv")
target_col = "CO2"
numeric_df = df.select_dtypes(include=[np.number]).copy()
all_nan_numeric_cols = [c for c in numeric_df.columns if numeric_df[c].isna().all()]
if all_nan_numeric_cols:
    numeric_df = numeric_df.drop(columns=all_nan_numeric_cols)
X = numeric_df.drop(columns=[target_col])
y = numeric_df[target_col]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21)
imputer = SimpleImputer(strategy="median")
X_train_imp = imputer.fit_transform(X_train)
X_test_imp  = imputer.transform(X_test)
print(X_train_imp)
lr_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())])
lr_cv = cross_val_score(
    lr_pipe, X_train_imp, y_train, cv=10, scoring="neg_root_mean_squared_error")
lr_rmse_cv_mean = -lr_cv.mean()
lr_pipe.fit(X_train_imp, y_train)
lr_pred = lr_pipe.predict(X_test_imp)
lr_rmse_test = np.sqrt(mean_squared_error(y_test, lr_pred))
print("\n[LR]   CV RMSE (mean): %.4f" % lr_rmse_cv_mean)
print("[LR]   Test RMSE:      %.4f" % lr_rmse_test)

```
