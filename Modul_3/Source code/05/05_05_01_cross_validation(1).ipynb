```python
from sklearn import datasets, neighbors, metrics, model_selection as skms
diabetes = datasets.load_diabetes()

# data, model, fit & cv-score
model = neighbors.KNeighborsRegressor(10)
scores = skms.cross_val_score(model,
  diabetes.data,
  diabetes.target,
  cv=5,
  scoring='neg_mean_squared_error')
# zadano za cross_val_score:
# cv=3 fold, no shuffle, stratified if classifier
# model.score zadano (regresor: r2, klasifikator: accuracy)
print(scores)
print(scores.mean())
print(scores.std())
```

```python
import matplotlib.pyplot as plt

# Pretvorba negativnog MSE u pozitivni MSE
mse_scores = -scores

# Kategorije za stupce
folds = [f"Fold {i+1}" for i in range(len(mse_scores))]

# Crtanje stupčastog grafikona
plt.bar(folds, mse_scores, color='skyblue')
plt.ylabel("MSE")
plt.title("Rezultati MSE po foldovima (KNN)")
plt.show()
```
