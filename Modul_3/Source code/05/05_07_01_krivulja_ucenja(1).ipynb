```python
from sklearn import datasets
from sklearn import neighbors
from sklearn import model_selection as skms
import numpy as np
import pandas as pd

iris = datasets.load_iris()
# 10 podatkovnih skupova velicine: 10% - 100%
# (5-fold CV)
train_sizes = np.linspace(.1, 1.0, 10)
nn = neighbors.KNeighborsClassifier()
(train_N, train_scores, test_scores) = skms.learning_curve(nn,
  iris.data, iris.target,
  cv=5, train_sizes=train_sizes)
# sažeti kroz 5 CV rezultata; jedan rezultat za svaku veličinu skupa podataka
df = pd.DataFrame(test_scores, index=(train_sizes*100).astype(int))
df['Mean 5-CV'] = df.mean(axis='columns')
df.index.name = "% Data Used"
display(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# kreiraj dataframe iz rezultata
train_df = pd.DataFrame(train_scores)
test_df = pd.DataFrame(test_scores)

# dodaj stupac za veličine skupova za treniranje
train_df['Data Used (%)'] = train_sizes * 100
test_df['Data Used (%)'] = train_sizes * 100

# dodaj stupac Train/Test
train_df['Type'] = 'Train'
test_df['Type'] = 'Test'

# Pretvori DataFrames u dugi format
train_melted = train_df.melt(id_vars=['Data Used (%)', 'Type'], var_name='Fold', value_name='Accuracy')
test_melted = test_df.melt(id_vars=['Data Used (%)', 'Type'], var_name='Fold', value_name='Accuracy')

# Spojite DataFrameove
combined_df = pd.concat([train_melted, test_melted])

# Crtanje sa lineplot
plt.figure(figsize=(10, 6))
ax = sns.lineplot(data=combined_df, x='Data Used (%)', y='Accuracy', hue='Type')

ax.set_title("Krivulja učenja 5-NN klasifikator")
ax.set_xlabel("Postotak podataka za treniranje")
ax.set_ylabel("Točnost")
plt.grid(True)
plt.show()
```

```python
num_neigh = [1, 3, 5, 10, 15, 20]
KNC = neighbors.KNeighborsClassifier
tt = skms.validation_curve(KNC(),
  iris.data, iris.target,
  param_name='n_neighbors',
  param_range=num_neigh,
  cv=5)

# tt je n-torka od (rezultati_vlaka, rezultati_testova)
train_scores, test_scores = tt

# Preoblikujte podatke za linijski dijagram
# Izradite okvire podataka iz rezultata
train_df = pd.DataFrame(train_scores)
test_df = pd.DataFrame(test_scores)

# Dodajte stupac za vrijednosti parametara
train_df['n_neighbors'] = num_neigh
test_df['n_neighbors'] = num_neigh

# Dodajte stupac za tip (Train ili Test)
train_df['Type'] = 'Train'
test_df['Type'] = 'Test'

# Pretvori DataFrameove u dugi format
train_melted = train_df.melt(id_vars=['n_neighbors', 'Type'], var_name='Fold', value_name='Accuracy')
test_melted = test_df.melt(id_vars=['n_neighbors', 'Type'], var_name='Fold', value_name='Accuracy')

# Spoji  DataFrameove
combined_df = pd.concat([train_melted, test_melted])

# Crtanje pomoću linijskog dijagrama
plt.figure(figsize=(10, 6))
ax = sns.lineplot(data=combined_df, x='n_neighbors', y='Accuracy', hue='Type')


ax.set_title('5-fold CV performanse za k-NN')
ax.set_xlabel("\n".join(['k for k-NN',
  'manji k, kompleksnije',
  'veći k, manje kompleksno']))
ax.set_ylim(.9, 1.01)
ax.set_ylabel('Accuracy');
plt.grid(True)
plt.show()
```
