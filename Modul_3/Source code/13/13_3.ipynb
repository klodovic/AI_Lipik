```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

X, y = load_iris(return_X_y=True)
target_names = ["setosa", "versicolor", "virginica"]

X_std = StandardScaler().fit_transform(X)

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_std)

print("Postotak varijance po komponenti:", pca.explained_variance_ratio_)
print("Kumulativno objašnjena varijanca:", pca.explained_variance_ratio_.sum())

plt.figure(figsize=(6,5))
for i, name in enumerate(target_names):
    plt.scatter(X_pca[y==i, 0], X_pca[y==i, 1], label=name, s=40)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Iris — PCA (2D)")
plt.legend()
plt.tight_layout()
plt.show()
```
