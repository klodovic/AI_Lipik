```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import numpy as np
#treba se poigrati sa random_state=1,21,42,43
X, y = make_blobs(n_samples=400, centers=4, random_state=43, cluster_std=1.2)

inertia = []  # spremnik za rezultate (unutarnja suma kvadrata)
K = range(1, 10)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(7, 5))
plt.plot(K, inertia, 'bo-')
plt.title('Metoda lakta za pronalaženje optimalnog K')
plt.xlabel('Broj klastera (K)')
plt.ylabel('Inercija (zbroj kvadratnih udaljenosti)')
plt.grid(True)
plt.show()

#ovo mijenjamo od 1 do optimalnog, a možemo i preko toga za probu
optimal_k = 1
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
kmeans.fit(X)
labels = kmeans.labels_
centers = kmeans.cluster_centers_

plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, s=50, cmap='viridis', alpha=0.7)
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, marker='X', label='Centroidi')
plt.title(f"K-means klasteriranje (K = {optimal_k})")
plt.xlabel("Značajka 1")
plt.ylabel("Značajka 2")
plt.legend()
plt.grid(True)
plt.show()
```
